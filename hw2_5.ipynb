```python
import numpy as np

n = int(input("Введіть довжину масивів: "))

arr1 = np.random.randint(0, 11, n)
arr2 = np.random.randint(10, 21, n)

print("Перший масив:", arr1)
print("Другий масив:", arr2)

combined = np.concatenate((arr1, arr2))
print("Об'єднаний масив:", combined)

sum_arr = arr1 + arr2
print("Сума масивів:", sum_arr)

diff_arr = arr1 - arr2
print("Різниця масивів:", diff_arr)
```
