```python
import numpy as np
arr = np.random.randint(0, 51, 20)
print("Початковий масив:", arr)
porih = int(input("Введіть порогове значення: "))

quantity = np.sum(arr > porih)
print("Кількість елементів, більших за поріг:", quantity)

max_value = np.max(arr)
position = np.argmax(arr)

print("Максимальне значення:", max_value)
print("Позиція першого максимального елемента:", position)

sorted_arr = np.sort(arr)[::-1]
print("Масив за спаданням:", sorted_arr)
```
